```python
from pyspark.sql import functions as F
```

```python
fact_transactions= spark.table("workspace.default.fact_transactions")
```

```python
row_count = fact_transactions.count()
distinct_transaction_count = (fact_transactions.select("transaction_id").distinct().count())
```

```python
null_key_count = fact_transactions.filter(F.col("customer_key").isNull()|F.col("card_key").isNull()| F.col("merchant_key").isNull()|F.col("date_key").isNull()).count()
```

```python
print(f"Fact row count: {row_count}")
print(f"Distinct transaction IDs: {distinct_transaction_count}")
print(f"Rows with null dimension keys: {null_key_count}")
```

```python
# Fail the notebook if data-quality checks fail
assert row_count == distinct_transaction_count, \
    "Validation failed: duplicate transaction IDs found."
assert null_key_count == 0, \
    "Validation failed: null dimension keys found."

print("Gold validation passed.")
```
